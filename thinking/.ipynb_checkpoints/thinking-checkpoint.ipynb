```python
import json
```

```python
# Load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data("data.json")
print(data)
print(type(data))
```

```python
# clean and structure the data
def clean_data(data):
    text_to_numbers = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    clean_data = []
    unique_user = set() 
    for user in data:
        # clean rating
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in text_to_numbers):
            raw_rating = text_to_numbers[raw_rating]
        user["rating"] = raw_rating
 
        # handle missing values
        raw_age = user.get("age")
        if( raw_age == None):
            user["age"] = None

       # Remove Duplicate data     
        if( user["name"].strip() in unique_user): 
             continue
        unique_user.add(user["name"])   
        clean_data.append(user)

    return clean_data
```

```python
data = clean_data(data)
```

```python
# get meaningfull insights from data 
def get_insights(data):

    # avg rating
    total_rating = 0
    for user in data:
        total_rating += float(user["rating"])
    print(f"avg rating:{total_rating/len(data)}")   

# percentage of user with poor rating 

    poor_ratings = 0
    for user in data:
        if(float( user["rating"]) < 3):
            poor_ratings += 1
    print(f"% of user with poor rating {poor_ratings/len(data) * 100}%")        
```

```python
get_insights(data)
```

```python
# build the recommendtation feataure  
def recommendations(data):
    recommendations = []

    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]
        if(float(user["rating"]) >= 4):
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"
        recommendations.append(curr_recomm)             
    return recommendations        
```

```python
recommendations(data)
```
